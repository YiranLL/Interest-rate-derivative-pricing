#### Key Idea:
**Discount factor**: $P(0,T)$ is the value today of receiving one unit of currency at time $T$. Mathematically, it is defined as $$P(0,T) = \frac{PV}{FV}$$, and if we are considering continuous compounding, then $P(0,T) = e^{-rT}$.

Notice that if the interest rate is positive, then $P(0,T)$ lies between $(0,1]$.


**If Annual Compounding**: £1 today worths $(1+r)^T$ at time T (years). So discount factor is $(1+r)^{-T}$.

**Zero rate**: the interest rate for a single cash flow paid at one specific future date, with no intermediate payments. Mathematically, $$z(T) = -\frac{\log P(0,T)}{T}$$.


**Growth factor**: Suppose we are looking for the impled growth of £1 today from time $T_1$ to time $T_2$, then growth rate is defined as $$\frac{P(0,T_1)}{P(0,T_2)}$$. 

**Forward rate**: It is defined as the fractional part of growth factor, mathematically, $$ F(T_1,T_2) = \frac{P(0,T_1) - P(0,T_2)}{\alpha  P(0,T_2)}$$

**Cashflow**: take the annual interest amount $N\times r$, then scale it by the fraction of a year represented by this coupon period, with $$CF = N \times \alpha \times r$$, where   
$N$ = notional, the principal amount the interest is based on,   
$r$ = annualised interest rate,  
$\alpha$ = year fraction for this accrual period.  

**Interest rate swap**: Swap fixed rate interest and floating rate interest.

#### Yield Curve Bootstrapping
Yield curve describes the interest rate and accural period. There are three relevant curves.  
1. **Zero curve**: $z(T)$ is the zero interest rate for different accural period.
2. **Discount curve**: $P(0,T)$ is the discounted factor for different future dates.
3. **Forward curve**: $F(T_1, T_2)$ is the implied future interest rate from today.
  
The three curves are not independent, given discounted curve, we can calculate the remaining curves.

##### Why is bootstrapping necessary?

Normally there are only a mixture of different instruments and maturities, they do not use the same type of interest rate. Hence we use bootstrapping to convert these market quotes of different instruments to a sequence of discount factors.
  
**Main Idea**: Every time we choose an instrument that has a farther maturity, we use the discount factors calculated from the past, to calculate a new discount factor.

##### Deposit
Suppose today we save $N$, mature at $T$, simple annual rate $R$, accural fraction $\alpha$. At maturity, the return should be $N(1+R\alpha)$, by no arbitrage principle, we must also have $N = N(1+R\alpha) P(0,T)$. Then $$P(0,T) = \frac{1}{1+R\alpha}$$.

##### FRA
FRA always quotes for a forward rate, so use deposit and FRA we can calculate $P(0,T)$ for a $T$ in future.


##### Interest Rate Swap
We need swap quotes to construct yield curve over a longer accural period. At the moment the contract is set up, value of the swap is 0. Assume the par swap rate is $K$.  
1. PV of fixed leg at par is $$PV_{fixed} = NK \sum^{n}_{i=1} \alpha_i P(0,T_i)$$.
So we define the **swap annuity** to be $$A = \sum^n_{i=1} \alpha_i P(0,T_i)$$    
2. PV of floating leg at par is $$PV_{floating} = N(1-P(0,T_n))$$  
    
    
At par, $PV_{fixed} = PV_{floating}$. Hence we can derive the **par swap rate formula** $$K = \frac{1-P(0,T_n)}{\sum^n_{i=1} \alpha_i P(0,T_i)}$$. Hence if we have par swap rate from the market quote, we can derive a discount curve.
